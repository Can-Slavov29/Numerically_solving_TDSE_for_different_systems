# Solving schrodingers equation using finite differenced method

# Infinite square well

```python
#import relevant modules
import numpy as np
import matplotlib.pyplot as plt
import scipy as sci

#set paramters and constants
m = 9.1e-31
N = 4000
a = 0.1e-9
h = 6.626e-34
h_bar = h/(2*np.pi)
dx = a/(N+1)

#setting up spatial grid
x_values = np.linspace(dx,a-dx,N)


#creating matrix to solve
coef = 1/(dx**2) * -(h_bar)**2/(2*m)
H =coef * (np.diag(np.full(N,-2)) + np.diag(np.ones(N-1),1) + np.diag(np.ones(N-1),-1))



energies,vectors = sci.linalg.eigh(H)


def plot_wavefunction(n):
    if n==1 :
        suffix = "'st"
    elif n==2:
        suffix = "'nd"
    elif n==3:
        suffix = "'rd"
    else:
        suffix = "'th"
    
    plt.figure()
    plt.title(f"{n}{suffix} Wavefunction")
    plt.xlabel("x values")
    plt.ylabel(r"$\Psi_{n}(x)$")
    plt.plot(x_values,vectors[:,n])


```

```python
plot_wavefunction(2)
```
